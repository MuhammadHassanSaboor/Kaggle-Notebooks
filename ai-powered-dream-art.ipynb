# **🌙 AI-Powered Dream / Story Visualization 🌌**

## **💭 Welcome!**  

Ever wondered what your dreams or imaginative stories would look like as real images?  
This notebook uses **cutting-edge AI** to transform **text prompts** into **stunning visuals** using **Stable Diffusion**, a state-of-the-art **text-to-image generative model**.  

**✨ What you will see here:**  
- Automatic generation of **dream-like images** from creative prompts 🖼️  
- Exploration of **generative AI in art and storytelling** 🤖  
- Interactive and visually stunning results that bring imagination to life 🌟  

**🚀 Let’s dive in and turn dreams into art!**

```python
# ===============================
# AI‑Powered Dream / Story to Image
# Automated Prompts Version
# ===============================

import torch
from diffusers import StableDiffusionPipeline
import matplotlib.pyplot as plt
from PIL import Image
import os

# 1) Check for GPU
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", DEVICE)

# 2) Load Stable Diffusion Model
# You can also use "stabilityai/stable-diffusion-2-1" or "runwayml/stable-diffusion-v1-5"
model_id = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16 if DEVICE=="cuda" else torch.float32,
).to(DEVICE)

# 3) Predefined prompts (dream/story ideas)
prompts = [
    "A floating city at sunrise, drawn in watercolor",
    "A robot reading poetry by candlelight",
    "A magical forest with glowing plants and creatures",
    "A futuristic cityscape at night, neon lights reflecting on water",
    "A dragon flying above a mountain range during sunset"
]

# 4) Folder to save images
output_folder = "dream_images"
os.makedirs(output_folder, exist_ok=True)

# 5) Generate images automatically
for idx, prompt in enumerate(prompts, 1):
    print(f"Generating image {idx}/{len(prompts)} for prompt: {prompt}")
    result = pipe(prompt, guidance_scale=7.5, num_inference_steps=50)
    image = result.images[0]
    
    # Display image
    image.show()
    
    # Save image
    filename = os.path.join(output_folder, f"dream_image_{idx}.png")
    image.save(filename)
    print(f"Saved image: {filename}")

print("All images generated and saved successfully!")
```

```python
plt.figure(figsize=(20, 10))
for idx, prompt in enumerate(prompts):
    img_path = os.path.join(output_folder, f"dream_image_{idx+1}.png")
    img = Image.open(img_path)
    plt.subplot(1, len(prompts), idx+1)
    plt.imshow(img)
    plt.title(f"Dream {idx+1}")
    plt.axis('off')
plt.show()
```
